NumPy es la librería líder para procesamiento numérico en Python.
- Los arrays de NumPy (`np.array`) son más rápidos y eficientes que las listas porque almacenan datos del mismo tipo en memoria contigua.
- Usamos .shape para ver laforma (filas, columnas) y .ndim para saber el numero de dimensiones

```python
import numpy as np

edades_lista = [10,34,23,65,78]

# Array o arreglo 
edades_array = np.array(edades_lista)

print("Vector de edades:", edades_array)
print("Tipo de objeto:", type(edades_array))
print("Forma (shape):" , edades_array.shape)
print("Dimensiones (ndim):", edades_array.ndim)
```

```python
millon_lista = list(range(1000000))
millon_array = np.array(millon_lista)

import time

t_inicio = time.time()
resultado_for = [x + 10 for x in millon_lista]
t_for = time.time() - t_inicio

print(f"Tiempo con bucle for: {t_for:.5f} segundos")

t_inicio_np = time.time()
resultado_np = millon_array + 10
tp_np = time.time() - t_inicio_np

print(f"Tiempo con NumPy vectorizado: {tp_np:.5f} segundos")
print(f"NumPy es aproximadamente {t_for/tp_np:.1f} veces más rapido")

```

## 3. Indexación 2D, Máscaras Booleanas y Estadística 

```python
#1. Crear matriz de 2 filas y 3 columnas 
ventas = np.array([
#horas 0 , 1, 2
    [120,150,90], # lunes [0]
    [200, 80, 250] #martes [1]
])

print(f"Matriz de ventas: {ventas}")
print(f"Ventas del día 1, hora 2 (fila 0, columna 1): {ventas[0,1]}")

#2. Filtrado boleano True / False 
ventas_altas = ventas[ventas > 100] 
print(f"Ventas mayores a 100:", ventas_altas)

#3. Metricas Estadísticas
#Promedio o media
print("Promedio general de ventas:", np.mean(ventas))
print(f"Mediana de ventas: {np.median(ventas)}")
print(f"Mediana de ventas: {np.median(ventas)}")
print(f"Desviacion estandar:{np.std(ventas)}")
print(f"Maxima venta registrada:{np.max(ventas)}")
```
