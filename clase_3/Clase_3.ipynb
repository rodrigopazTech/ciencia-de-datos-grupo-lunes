Métodos de texto y conversión de datos

.upper() y .lower() cambian mayúsculas/minúsculas.

.strip() quita espacios extras en los bordes.

.replace() sustituye texto.

```python
producto = " Laptop Pro 15 "
producto_limpios = producto.strip().upper()
print(f"Producto limpio en mayusculas: {producto_limpios}")

edad_ingresada = input("Por favor ingresa tu edad:")
edad_numero = int(edad_ingresada)

print(f"Texto ingresado: '{edad_ingresada}' Tipo: {type(edad_ingresada)}")
print(f"Numero Convertido '{edad_numero}' Tipo: {type(edad_numero)}")
print(f"En 5 años tendrás: {edad_numero + 5} años")
```

Extracción avanzada en listas:
Slicing y Métodos útiles

```python
venta_semana = [100, 150, 200, 80, 250, 300, 190]

# Extraer los primero 4 dias
primeros_dias = venta_semana[0:4]

# Extraer los últimos 2 días
# Numeros negativos
ultimos_dias = venta_semana[-2:]

# Extraer días intermedios (del indice 2 al 4)
dias_centrales = venta_semana[2:5]

print(f"Ventas completas: {venta_semana}")
print(f"Primeros 4 dias: {primeros_dias}")
print(f"Ultimas 2 ventas: {ultimos_dias}")
print(f"Dias centrales (2:5): {dias_centrales}")
```

Verificación de pertenencia y ordenamiento

```python
regiones = ["Norte", "Sur", "Centro", "Este"]

# Verificar si una región existe
print("¿Existe 'Sur' en regiones?:", "Sur" in regiones)

# Ordenar listas
precios = [45.5, 12.0, 99.9, 23.4]
precios.sort()
print("PRECIOS ORDENADOS DE MENOR A MAYOR:", precios)

# Eliminar el último precio
precio_removido = precios.pop()
print("Precio eliminado: ", precio_removido)
print("Lista resultante de precios:", precios)
```

###Comparaciones y Lógica

Comparadores: ==, !=, >, <, >=, <=.

Operadores lógicos:

and: verdadero si se cumplen ambas condiciones.

or: verdadero si se cumple al menos una condición.

not: invierte la condición.

```python
temperatura = 28.5
humedad = 65

# Mayor que
hace_calor = temperatura > 25

# Mayor o Igual que
humedo = humedad >= 60

print("¿Hace calor?", hace_calor)
print("¿Hace calor Y esta humedo?:", hace_calor and humedo)
print("¿Hace calor Ó la humedad es baja?:", hace_calor or humedad < 50)
print("¿Humedad es diferente a 60? ", humedad != 60)
```

Estructura condicional if / elif / else

```python
score_credito = 680

if score_credito >= 750:
    nivel_riesgo = "Bajo"
    aprobado = True
elif score_credito >= 650:
    nivel_riesgo = "Medio"
    aprobado = True
else:
    nivel_riesgo = "Alto"
    aprobado = False

print(f"Resultado de evaluación:")
print(f"Score: {score_credito} | Riesgo: {nivel_riesgo} | Crédito aprobado: {aprobado}")
```
